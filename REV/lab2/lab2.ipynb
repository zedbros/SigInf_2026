```python
import pandas as pd

df1 = pd.read_csv('river-A.csv')
df2 = pd.read_csv("river-B.csv")

a_debit = 15
b_debit = 21

r = max(a_debit*len(df1), b_debit*len(df2))

signal = []

for i in range(r):
    a = 0
    b = 0
    if i % a_debit == 0:
        a = df1.iloc[i % a_debit].metersPerSecond
    if i % b_debit == 0:
        b = df2.iloc[i % b_debit].metersPerSecond
    signal.append(a+b)

print(signal)
print(len(signal))
print(signal[315])
```

# Partie 4

```python
T = 10
dt = 0.1
f = 1.0

i_T = int(T/dt)

l = [i/10 for i in range(0, i_T)]

imp = [1 if i == 5.0 else 0 for i in l]
ech = [1 if i >= 2 else 0 for i in l]
sin = [math.sin(2*math.pi*f*i) for i in l]

imp_E = sum(pow(i, 2)*dt for i in imp)
ech_E = sum(pow(i, 2)*dt for i in ech)
sin_E = sum(pow(i, 2)*dt for i in sin)
print(f"Energie of all:\nImpulse {imp_E}\nEchelonne {ech_E}\nsin {sin_E}")

imp_P = imp_E/T
ech_P = ech_E/T
sin_P = sin_E/T
print(f"\nPower of all:\nImpulse {imp_P}\nEchelonne {ech_P}\nsin {sin_P}")
```

Impulse energy will stay the same but the power will infitesimaly reach 0 since it is a finite signal.

For the sin, the energy will increase but since it is a infinite repeating signal, well it stays the same.
